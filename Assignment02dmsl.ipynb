```python
#Visual representation 
import matplotlib.pyplot as plt
```

```python
#line graph 
plt.figure(figsize=(8, 5))
plt.plot(df["age"].head(20), marker="o")
plt.title("Age of First 20 Passengers")
plt.xlabel("Passenger Index")
plt.ylabel("Age")
plt.show()
```

```python

# Bar chart - Number of passengers by class

class_counts = df["pclass"].value_counts().sort_index()
plt.figure(figsize=(8, 5))
plt.bar(class_counts.index, class_counts.values, edgecolor="black")
plt.title("Number of Passengers by Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")
plt.xticks([1, 2, 3])
plt.show()
```

```python

plt.figure(figsize=(8, 5))
plt.hist(df["age"].dropna(), bins=15, edgecolor="black")
plt.title("Distribution of Passenger Ages")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
#scatter plot 
plt.figure(figsize=(8, 5))
plt.scatter(df["age"], df["fare"], alpha=0.5)
plt.title("Passenger Age vs Ticket Fare")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()
```

```python
#pie chart 
survival_counts = df["survived"].value_counts().sort_index()
plt.figure(figsize=(6, 6))
plt.pie(
    survival_counts.values,
    labels=["Did Not Survive", "Survived"],
    autopct="%1.1f%%"
)
plt.title("Titanic Passenger Survival")
plt.show()
```
