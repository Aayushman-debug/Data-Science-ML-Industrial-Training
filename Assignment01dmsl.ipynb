```python

import numpy as np
import pandas as pd

# Library Management System

class Library:
    def __init__(self):
        self.books = pd.DataFrame({
            "Book_ID": [101, 102, 103, 104, 105],
            "Title": [
                "Thinking, Fast and Slow",
                "The Art of War",
                "The Republic",
                "Bobby Fischer Teaches Chess",
                "The Psychology of Money"
            ],
            "Author": [
                "Daniel Kahneman",
                "Sun Tzu",
                "Plato",
                "Bobby Fischer",
                "Morgan Housel"
            ],
            "Copies": [3, 2, 2, 4, 3]
        })

        self.issued = []

    # Display all books
    def show_books(self):
        print("\nAvailable Books:")
        print(self.books.to_string(index=False))

    # Add a book
    def add_book(self):
        try:
            book_id = int(input("Enter book ID: "))

            if book_id in self.books["Book_ID"].values:
                print("Book ID already exists.")
                return

            title = input("Enter book title: ")
            author = input("Enter author name: ")
            copies = int(input("Enter number of copies: "))

            if title == "" or author == "" or copies < 0:
                print("Please enter valid details.")
                return

            new_book = pd.DataFrame([{
                "Book_ID": book_id,
                "Title": title,
                "Author": author,
                "Copies": copies
            }])

            self.books = pd.concat(
                [self.books, new_book],
                ignore_index=True
            )

            print("Book added successfully.")

        except ValueError:
            print("Please enter valid numbers.")

    # Search for a book
    def search_book(self):
        name = input("Enter book title to search: ")

        result = self.books[
            self.books["Title"].str.contains(
                name, case=False, na=False
            )
        ]

        if result.empty:
            print("Book not found.")
        else:
            print(result.to_string(index=False))

    # Issue a book
    def issue_book(self):
        try:
            book_id = int(input("Enter book ID: "))
            student = input("Enter student name: ").strip()

            if student == "":
                print("Enter a student name.")
                return

            book = self.books[self.books["Book_ID"] == book_id]

            if book.empty:
                print("Book does not exist.")
                return

            index = book.index[0]

            if self.books.loc[index, "Copies"] > 0:
                self.books.loc[index, "Copies"] -= 1

                self.issued.append({
                    "Book_ID": book_id,
                    "Title": self.books.loc[index, "Title"],
                    "Student": student
                })

                print("Book issued successfully.")
            else:
                print("No copies available.")

        except ValueError:
            print("Enter a valid book ID.")

    # Return a book
    def return_book(self):
        try:
            book_id = int(input("Enter book ID: "))
            student = input("Enter student name: ").strip()

            for i in range(len(self.issued)):
                book = self.issued[i]

                if (
                    book["Book_ID"] == book_id
                    and book["Student"].lower() == student.lower()
                ):
                    self.issued.pop(i)

                    index = self.books.index[
                        self.books["Book_ID"] == book_id
                    ][0]

                    self.books.loc[index, "Copies"] += 1

                    print("Book returned successfully.")
                    return

            print("No matching issued book found.")

        except ValueError:
            print("Enter a valid book ID.")

    #issued books
    def show_issued(self):
        if len(self.issued) == 0:
            print("No books have been issued.")
        else:
            issued_df = pd.DataFrame(self.issued)
            print(issued_df.to_string(index=False))

    # Basic stats
    def statistics(self):
        copies = self.books["Copies"].to_numpy()

        print("\nLibrary Statistics")
        print("Total books:", np.sum(copies))
        print("Number of different titles:", len(self.books))
        print("Average copies:", round(np.mean(copies), 2))
        print("Maximum copies:", np.max(copies))
        print("Minimum copies:", np.min(copies))


# Main program
library = Library()

while True:
    print("\n===== Library Management System =====")
    print("1. Show all books")
    print("2. Add a book")
    print("3. Search for a book")
    print("4. Issue a book")
    print("5. Return a book")
    print("6. Show issued books")
    print("7. Show statistics")
    print("8. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        library.show_books()

    elif choice == "2":
        library.add_book()

    elif choice == "3":
        library.search_book()

    elif choice == "4":
        library.issue_book()

    elif choice == "5":
        library.return_book()

    elif choice == "6":
        library.show_issued()

    elif choice == "7":
        library.statistics()

    elif choice == "8":
        print("Thank you for using the library system.")
        break

    else:
        print("Invalid choice. Try again.")
```
